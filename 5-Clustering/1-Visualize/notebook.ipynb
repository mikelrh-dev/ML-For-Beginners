# Clustering: Visualización de datos de música nigeriana

En esta lección vamos a explorar un dataset de música nigeriana de Spotify para entender la estructura de los datos antes de aplicar clustering.

**¿Qué es clustering?** Es un tipo de aprendizaje no supervisado que agrupa datos sin etiquetas según patrones que encuentra en ellos.

Imaginá que tenés una pila de ropa para ordenar 🧦👕👖🩲 — no tenés etiquetas, pero agrupás por similitud. Eso es clustering.

## 1. Cargar los datos

Cargamos el dataset de canciones nigerianas de Spotify. Tiene 530 canciones con 16 columnas: nombre, artista, género, popularidad, bailabilidad, energía, etc.

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("../data/nigerian-songs.csv")
df.head()
```

## 2. Explorar la estructura

`info()` nos muestra: 530 canciones, 16 columnas, tipos de datos, y que no hay valores nulos.

```python
df.info()
```

## 3. Verificar valores nulos

Si todos son 0, el dataset está limpio.

```python
df.isnull().sum()
```

## 4. Estadísticas descriptivas

`describe()` nos da la distribución de cada columna numérica. Notá que la popularidad puede ser 0 — esas son canciones sin clasificar.

```python
df.describe()
```

## 5. Géneros más populares

¿Cuáles son los géneros con más canciones en el dataset?

```python
import seaborn as sns

top = df['artist_top_genre'].value_counts()
plt.figure(figsize=(10,7))
sns.barplot(x=top[:5].index,y=top[:5].values)
plt.xticks(rotation=45)
plt.title('Top genres',color = 'blue')
```

## 6. Limpiar datos faltantes

Los géneros clasificados como 'Missing' no tienen clasificación de Spotify. Los eliminamos.

```python
df = df[df['artist_top_genre'] != 'Missing']
top = df['artist_top_genre'].value_counts()
plt.figure(figsize=(10,7))
sns.barplot(x=top.index,y=top.values)
plt.xticks(rotation=45)
plt.title('Top genres',color = 'blue')
```

## 7. Filtrar los 3 géneros principales

Nos quedamos con afro dancehall, afropop y nigerian pop. También eliminamos canciones con popularidad 0 (ruido).

```python
df = df[(df['artist_top_genre'] == 'afro dancehall') | (df['artist_top_genre'] == 'afropop') | (df['artist_top_genre'] == 'nigerian pop')]
df = df[(df['popularity'] > 0)]
top = df['artist_top_genre'].value_counts()
plt.figure(figsize=(10,7))
sns.barplot(x=top.index,y=top.values)
plt.xticks(rotation=45)
plt.title('Top genres',color = 'blue')
```

## 8. Matriz de correlación

¿Qué variables están relacionadas? El heatmap muestra correlaciones entre todas las columnas numéricas.

**Resultado:** La correlación más fuerte es entre `energy` y `loudness` (la música fuerte suele ser energética). El resto son débiles — ¡buen candidato para clustering!

```python
corrmat = df.corr(numeric_only=True)
f, ax = plt.subplots(figsize=(12, 9))
sns.heatmap(corrmat, vmax=.8, square=True)
```

## 9. Distribución por género

Un gráfico KDE (Kernel Density Estimate) muestra la densidad de puntos. Los círculos concéntricos indican dónde se agrupan las canciones según popularidad y bailabilidad.

```python
sns.set_theme(style="ticks")

g = sns.jointplot(
    data=df,
    x="popularity", y="danceability", hue="artist_top_genre",
    kind="kde",
)
```

## 10. Scatter plot

Un scatter plot simple muestra lo mismo: los tres géneros se superponen bastante. Determinar clusters en estos datos va a ser un desafío interesante.

```python
sns.FacetGrid(df, hue="artist_top_genre", height=5) \
   .map(plt.scatter, "popularity", "danceability") \
   .add_legend()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model, model_selection

# Load the diabetes dataset
X, y = datasets.load_diabetes(return_X_y=True)

# Print the shape of the data and the first row
print(X.shape)
print(X[0])
```

```python
# Nos quedamos solo con la columna del índice de masa corporal (BMI)
X = X[:, 2]
print("Forma tras aislar la columna:", X.shape)

# Lo transformamos en una matriz de dos dimensiones (2D)
X = X.reshape((-1, 1))
print("Forma final requerida por Scikit-learn:", X.shape)
```

```python
# Dividimos de forma aleatoria: 67% para entrenar y 33% para el examen final
X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.33, random_state=42)

print("Pacientes totales en X:", X.shape[0])
print("Pacientes para entrenar (X_train):", X_train.shape[0])
print("Pacientes para el examen (X_test):", X_test.shape[0])
```

```python
# 1. Creamos un objeto vacío que contiene el algoritmo matemático de Regresión Lineal
model = linear_model.LinearRegression()

# 2. Le pasamos los apuntes de entrenamiento para que busque la correlación
model.fit(X_train, y_train)

print("¡Modelo entrenado con éxito!")
```

```python
# Le pasamos los pesos de los 146 pacientes que NUNCA ha visto para que adivine su progresión
y_pred = model.predict(X_test)

print("Forma del array de predicciones:", y_pred.shape)
print("Primera predicción calculada por la IA:", y_pred[0])
```

```python
# 1. Pintamos los datos reales del examen (puntos negros independientes)
plt.scatter(X_test, y_test, color='black')

# 2. Pintamos la línea recta azul que representa las predicciones de nuestra IA
plt.plot(X_test, y_pred, color='blue', linewidth=3)

# 3. Añadimos las etiquetas de los ejes y el título para que sea legible
plt.xlabel('Scaled BMIs')
plt.ylabel('Disease Progression')
plt.title('Progression of Diabetes Against BMI')

# 4. Forzamos a Matplotlib a renderizar y mostrar la ventana del gráfico
plt.show()
```

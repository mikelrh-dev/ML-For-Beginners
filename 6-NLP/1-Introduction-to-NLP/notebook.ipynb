## 1. ¿Qué es NLP?

## 2. El Test de Turing

En los años 50, **Alan Turing** se preguntó: *"¿Puede una computadora pensar?"*

Propuso una prueba:
1. Un humano conversa por texto con alguien (sin verlo)
2. Puede ser otro humano O una computadora
3. Si después de un rato el humano no puede distinguir con quién habla → la computadora "piensa"

### El juego de la imitación
La idea vino de un juego de fiesta: un interrogador debe determinar quién es hombre y quién es mujer entre dos personas en otra habitación, solo con texto.

### ¿Por qué importa?
Es la base de toda la investigación en IA conversacional. Si una computadora puede engañar a un humano, algo está haciendo bien.

## 3. Eliza — El primer chatbot (1960s)

**Joseph Weizenbaum** del MIT creó Eliza, una computadora "terapeuta" que:

1. Hacía preguntas al usuario
2. Daba la apariencia de entender las respuestas
3. Respondía de forma coherente

### ¿Cómo funcionaba?

Si le decías: **"Yo estoy triste"**
Eliza respondía: **"¿Cuánto tiempo llevas triste?"**

¿Qué hizo? Simple:
- Detectó "yo estoy" + un adjetivo
- Cambió el tiempo verbal ("estoy" → "llevas")
- Reordenó la oración

**No entendía** — solo reorganizaba texto. Pero era suficiente para engañar a mucha gente.

### Ejemplo de falla
Si le decías: **"Tú eres una bicicleta"**
Eliza respondía: **"¿Cuánto tiempo he sido una bicicleta?"** 🚲

Obviamente no tenía sentido, pero mostraba los límites del enfoque.

## 4. Nuestro bot conversacional

Vamos a crear un bot **aún más simple que Eliza** — solo elige respuestas al azar.

### El plan
1. Mostrar instrucciones
2. Empezar un loop infinito
3. Aceptar entrada del usuario
4. Si escribe "bye" → salir
5. Si no → elegir respuesta aleatoria
6. Repetir

````python
import random

# Lista de respuestas aleatorias
random_responses = [
    "That is quite interesting, please tell me more.",
    "I see. Do go on.",
    "Why do you say that?",
    "Funny weather we've been having, isn't it?",
    "Let's change the subject.",
    "Did you catch the game last night?"
]

print("Hello, I am Marvin, the simple robot.")
print("You can end this conversation at any time by typing 'bye'")
print("After typing each answer, press 'enter'")
print("How are you today?")

while True:
    # Esperar que el usuario escriba algo
    user_input = input("> ")
    
    # Si escribió 'bye', salir del loop
    if user_input.lower() == "bye":
        break
    else:
        # Elegir una respuesta aleatoria de la lista
        response = random.choices(random_responses)[0]
    
    print(response)

print("It was nice talking to you, goodbye!")
````
